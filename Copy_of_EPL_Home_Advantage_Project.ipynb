### Do Premier League teams tend to do better at home, and has that changed over time?

## Project Overview
This project studies whether home advantage in the English Premier League changed between the 2000/01 and 2024/25 seasons. We will use match results and goals to measure how often home teams win and how many more goals they score than away teams.
We will also examine the 2020/21 season and compare it with nearby seasons, since matches were played under unusual crowd restrictions.







## Project Phases

1. **Data Description, Cleaning, and Exploratory Analysis**
learn what each row and column represents. Check for missing or suspicious values, then make simple summaries and plots to see patterns.

2. **Descriptive Analysis of Home Advantage**
Before we build mathematical models, we need to analyise and understand what happend:
   - What is the percentage of home wins? Away win? Draw?
   - Compare average home and away goals, goal difference, and how often each side scores or concedes.
   - Points earned per match at home and away?
   - Compare average yellow and red cards per match, and how often the home team receives more, fewer, or the same number as the away team.
   - Plot these measures season by season, then compare 2020/21 with the seasons just before and after it.

3. **Statistical Models and Pattern Analysis**
   - Massey ratings and the zero-sum constraint
   - Poisson model with maximum likelihood
   - PCA using SVD

4. **Match-Outcome Prediction**
   - Softmax regression using team ratings and recent form
   - Gradient descent, SGD, and ADAM

5. **Interpretation and Presentation**

| Project task | Method | Purpose |
|---|---|---|
| Estimate team strength and home advantage | Massey least squares | Calculates team ratings and the home-effect estimate, *h*. |
| Make the Massey ratings solvable | Rank/nullspace constraint | Sets the ratings to sum to zero. This is part of the Massey method, not a separate model. |
| Model goals | Poisson model with maximum likelihood estimation (MLE) | Estimates expected goals and the home effect. |
| Find patterns in match statistics | PCA using SVD | Summarizes related statistics, such as shots, corners, fouls, and cards. |
| Predict match outcomes | Softmax regression with gradient descent, SGD, and ADAM | Predicts home win, draw, or away win from team ratings and recent form. |

# Phase 3A — Massey Method
## Team Strength Ratings Using Least Squares

```python
import pandas as pd

df = pd.read_csv("epl_final.csv")

print(df.shape)
print(df.columns.tolist())
```

```python

# Select the columns needed for the Massey Method
massey_data = df[
    ["Season", "HomeTeam", "AwayTeam",
     "FullTimeHomeGoals", "FullTimeAwayGoals"]
].copy()

# Calculate goal difference
massey_data["GoalDifference"] = (
    massey_data["FullTimeHomeGoals"]
    - massey_data["FullTimeAwayGoals"]
)

# Display the first 5 matches
massey_data.head()
```

```python

# Store ratings from all seasons
all_ratings = []
results = []

for season_name, data in massey_data.groupby("Season"):

    # Identify teams
    season_teams = sorted(
        set(data["HomeTeam"]) | set(data["AwayTeam"])
    )

    team_index = {
        team: i for i, team in enumerate(season_teams)
    }

    # Build the Massey matrix
    M_season = np.zeros((len(data), len(season_teams)))

    for i, (_, match) in enumerate(data.iterrows()):
        M_season[i, team_index[match["HomeTeam"]]] = 1
        M_season[i, team_index[match["AwayTeam"]]] = -1

    # Calculate rank and nullity
    rank_M = np.linalg.matrix_rank(M_season)
    nullity_M = len(season_teams) - rank_M

    results.append({
        "Season": season_name,
        "Matches": len(data),
        "Teams": len(season_teams),
        "Rank": rank_M,
        "Nullity": nullity_M
    })

    # Goal difference vector
    d_season = data["GoalDifference"].to_numpy()

    # Solve least squares
    r, _, _, _ = np.linalg.lstsq(
        M_season, d_season, rcond=None
    )

    # Make ratings sum to zero
    r = r - r.mean()

    # Save each team's rating
    for team, rating in zip(season_teams, r):
        all_ratings.append({
            "Season": season_name,
            "Team": team,
            "MasseyRating": rating
        })

# Combine results into one table
all_ratings_df = pd.DataFrame(all_ratings)

print("Number of seasons:",
      all_ratings_df["Season"].nunique())

print("Total team-season ratings:", len(all_ratings_df))

display(all_ratings_df.head())

verification_df = pd.DataFrame(results)
display(verification_df)
```

```python

# Add Massey ratings to each match
home_ratings = all_ratings_df.rename(columns={
    "Team": "HomeTeam",
    "MasseyRating": "HomeRating"
})

away_ratings = all_ratings_df.rename(columns={
    "Team": "AwayTeam",
    "MasseyRating": "AwayRating"
})

matches_rated = massey_data.merge(
    home_ratings, on=["Season", "HomeTeam"], how="left"
)

matches_rated = matches_rated.merge(
    away_ratings, on=["Season", "AwayTeam"], how="left"
)

# Expected goal difference based on team strengths
matches_rated["ExpectedGoalDifference"] = (
    matches_rated["HomeRating"] -
    matches_rated["AwayRating"]
)

# Actual performance minus expected performance
matches_rated["AdjustedHomeAdvantage"] = (
    matches_rated["GoalDifference"] -
    matches_rated["ExpectedGoalDifference"]
)

display(matches_rated.head())
```

```python

import matplotlib.pyplot as plt

# Average adjusted home advantage for each season
season_trend = (
    matches_rated.groupby("Season")["AdjustedHomeAdvantage"]
    .mean()
    .reset_index()
)

# Plot the trend
plt.figure(figsize=(12, 5))

plt.plot(
    season_trend["Season"],
    season_trend["AdjustedHomeAdvantage"],
    marker="o"
)

plt.axhline(0, linestyle="--", color="gray")

plt.xticks(rotation=90)
plt.xlabel("Season")
plt.ylabel("Average Residual Goal Difference (Goals)")
plt.title("Premier League Average Home Goal Difference Trend (2000/01–2024/25)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```


# Phase 3A — Massey Method: Team Strength Ratings

## 1. Objective

The objective of Phase 3A is to estimate the relative strengths of English Premier League teams using the **Massey Method**, which applies linear algebra and least-squares optimization.

We analyzed **25 seasons (2000/01–2024/25)** using **9,380 available matches**.

## 2. Constructing the Massey Matrix

For each season, we constructed a matrix $M$ where:

- Each row represents one match.
- Each column represents one team.
- The home team receives **+1**.
- The away team receives **−1**.
- All other teams receive **0**.

For example, the 2000/01 season has 380 matches and 20 teams:

$$
M \in \mathbb{R}^{380 \times 20}
$$

We calculated the goal difference for each match:

$$
d_i = \text{Home Goals}_i - \text{Away Goals}_i
$$

The goal-difference vector $d$ contains the results of all matches in the selected season.

## 3. Least-Squares Solution

The Massey Method estimates team-strength ratings by solving:

$$
Mr \approx d
$$

where:

- $M$ is the match matrix.
- $r$ is the vector of unknown team-strength ratings.
- $d$ is the goal-difference vector.

Because the system is generally overdetermined, we minimize the squared error:

$$
\min_r \|Mr-d\|_2^2
$$

The corresponding normal equations are:

$$
M^TMr=M^Td
$$

We used NumPy's `np.linalg.lstsq()` to calculate the ratings.

To make the ratings uniquely defined, we imposed the zero-sum constraint:

$$
\sum_{j=1}^{20}r_j=0
$$

This centers the team ratings around zero.

## 4. Rank and Nullspace Analysis

For every season, the Massey matrix had:

- **20 columns**
- **Rank = 19**
- **Nullity = 1**

Because each row contains one +1 and one −1, every row sums to zero.

Therefore:

$$
M\mathbf{1}=0
$$

where $\mathbf{1}$ is the vector containing twenty ones.

This shows that the all-ones vector belongs to the nullspace of $M$.

Using the rank-nullity theorem:

$$
\operatorname{rank}(M)+\operatorname{nullity}(M)=20
$$

$$
19+1=20
$$

Since the nullity is 1, the ratings are determined only up to an additive constant.

The zero-sum constraint removes this ambiguity and gives a unique least-squares solution.

## 6. Dataset Limitations

The dataset contains **9,380 matches**, compared with **9,500 expected matches** for 25 complete Premier League seasons.

| Season | Available Matches | Missing Matches |

We found **zero duplicate fixtures**.

The missing matches may affect the accuracy of team-strength ratings for those seasons.

Also, the ratings were calculated using all available matches within each season. Therefore, they describe historical team performance and should not be treated as leakage-free predictions of earlier matches.

## 7. Connection to Home Advantage

Our overall project investigates whether home advantage has declined over time.

The Massey Method helps estimate differences in team strength, which can be useful when interpreting match results.

For a match between a home team and an away team, the strength-based expected goal difference is:

$$
\widehat{d}_i=r_{\text{home}}-r_{\text{away}}
$$

The residual is:

$$
e_i=d_i-\widehat{d}_i
$$

A positive residual means the home team performed better than the strength-only model expected.

However, the basic Massey Method does not explicitly estimate a separate home-advantage effect. Additionally, the season-average residual does not independently establish whether home advantage is declining.

A separate home-advantage parameter or additional analysis would be needed to isolate that effect.
